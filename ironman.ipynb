# variable qualitative "division"

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("results.csv")
print(df.shape)
df.head()
```

```python

print(df["Division"].dtype)
print("Valeurs manquantes :", df["Division"].isna().sum())
print("Modalités distinctes :", df["Division"].nunique())
```

## Nettoyage
 On retire les valeurs manquantes, les chaînes vides et les espaces superflus.

```python
d = df["Division"].dropna().astype(str).str.strip()
d = d[d != ""]
print(f"Observations valides : {len(d)} sur {len(df)}")
print(f"Nombre de modalités : {d.nunique()}")
```

## Tableau des effectifs et fréquences 
 L’effectif est le nombre d’athlètes par division, la fréquence est ce nombre en pourcentage, et la fréquence cumulée montre combien de divisions couvrent l’essentiel des athlètes.

```python
eff = d.value_counts()
tab = pd.DataFrame({
    "Effectif": eff,
    "Frequence_%": (100 * eff / eff.sum()).round(2),
})
tab["Freq_cumulee_%"] = tab["Frequence_%"].cumsum().round(2)
print(tab.to_string())
```

## Le mode 

```python
print(f"Mode : {eff.index[0]} ({eff.iloc[0]} athlètes, "
      f"{100*eff.iloc[0]/eff.sum():.1f} %)")
```

Interprétation : Les hommes entre 40 et 44 ans participent souvent dans le triathlon .

## Graphiques

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 5))

eff.plot(kind="bar", ax=axes[0], color="steelblue")
axes[0].set_title("Effectifs par division")
axes[0].set_ylabel("Effectif")
axes[0].tick_params(axis="x", labelsize=7)

top = eff.head(8)
pie_data = pd.concat([top, pd.Series({"Autres": eff.iloc[8:].sum()})])
axes[1].pie(pie_data, labels=pie_data.index, autopct="%1.1f%%", startangle=90)
axes[1].set_title("Top 8 des divisions (+ Autres)")

plt.tight_layout()
plt.show()
```

## Interprétation des Graphiques

## Diagramme en barres :
 La division la plus représentée est M40-44, avec environ 176 000 athlètes. Viennent ensuite M35-39 (environ 155 000), M45-49 (environ 146 000), M30-34 (environ 122 000) et M50-54 (environ 102 000).
 Les effectifs décroissent rapidement : après ces 5 divisions, ils chutent sous 70 000 (M25-29), puis sous 50 000. La distribution est très concentrée sur quelques modalités.
 Les divisions féminines apparaissent plus loin dans le classement, avec des effectifs d’environ 30 000 à 40 000 (F40-44, F35-39, F30-34, etc.). Les hommes sont donc nettement plus nombreux que les femmes dans chaque tranche d’âge.
 À droite du graphique, il y a une très longue queue de modalités quasi invisibles (effectif proche de zéro). Plusieurs ne sont pas des catégories d’âge classiques (par exemple RELAY, MPRO, FPRO, FUNKNOWN, DNK), ce qui indique des modalités rares ou mal codées.

## Camembert (Top 8 + Autres)
 Les trois plus grandes parts sont M40-44 (16,1 %), M35-39 (14,1 %) et M45-49 (13,3 %).
 Les 7 premières divisions sont toutes masculines et représentent à elles seules 74,6 % des athlètes (16,1 + 14,1 + 13,3 + 11,1 + 9,3 + 6,1 + 4,6). La première division féminine, F40-44, n’arrive qu’en 8ᵉ position avec 3,5 %.
 La part « Autres » est de 21,9 % : elle regroupe toutes les divisions hors du top 8, y compris la majorité des divisions féminines et les modalités rares.

Indices de position

```python
print(f"Min : {x.min():.2f} | Max : {x.max():.2f}")
q1, q2, q3 = x.quantile([0.25, 0.50, 0.75])
print(f"Q1 : {q1:.2f} | Q2 : {q2:.2f} | Q3 : {q3:.2f}")
x.describe()
```

Indices de dispersion

```python
print(f"Variance : {x.var():.2f}")
print(f"Écart-type : {x.std():.2f}")
```

 Interprétation 
 L’écart-type (111,55 min) : un athlète finit environ 111 minutes (1 h 52) avant ou après la moyenne.
 Rapporté à la moyenne, cet écart représente environ 14,7 %, ce qui traduit une variabilité modérée autour de la moyenne .

Forme de la distribution

```python
skew_pearson = 3 * (x.mean() - x.median()) / x.std()
print("Indice d'asymétrie (Pearson) :", skew_pearson, "\n")
```

Distribution quasi symétrique

 Valeurs aberrantes

```python
iqr = q3 - q1
low, high = q1 - 1.5 * iqr, q3 + 1.5 * iqr
n_out = ((x < low) | (x > high)).sum()
print(f"Bornes : [{low:.1f} ; {high:.1f}]")
print(f"Valeurs aberrantes : {n_out} ({100 * n_out / len(x):.2f} %)")
```

Graphiques

Interprétation des Graphiques Univariés


       Boxplot (Boîte à moustaches) :
 La boîte, qui contient la moitié centrale des athlètes, s’étend environ de 680 à 850 minutes, avec la médiane (ligne orange) autour de 750 minutes, soit environ 12 h 30. La médiane est à peu près au centre de la boîte, ce qui indique une répartition équilibrée autour d’elle.
 Il y a des points isolés (outliers) aux deux extrémités : un groupe de temps très courts autour de 380 à 430 minutes, et quelques temps très longs, jusqu’à environ 1 400 minutes. Ces valeurs sont peu nombreuses par rapport à la taille du jeu de données.

  Histogramme (Distribution de fréquence) :
 L’histogramme montre une distribution unimodale (un seul pic), centrée autour de 700 à 750 minutes, avec une forme en cloche.
 La moyenne (ligne rouge) et la médiane (ligne noire) sont presque superposées, ce qui confirme une distribution quasi symétrique. La queue vers les temps élevés (800 à 1 000 minutes) est un peu plus étirée, ce qui correspond à une très légère asymétrie à droite, cohérente avec une moyenne légèrement supérieure à la médiane.
 Les outliers du boxplot (au-delà de 1 000 minutes) ne sont pas visibles sur l’histogramme, car ils sont trop rares.

  QQ-Plot (Quantile-Quantile Plot) :
 Les points suivent bien la droite rouge théorique dans la partie centrale, ce qui signifie que la majorité des temps se comporte comme une loi normale.
 Aux extrémités, les points s’écartent de la droite. À droite, ils se tassent autour de 1 000 à 1 030 minutes, au lieu de continuer à monter. C’est cohérent avec la limite de temps officielle d’un Ironman (17 heures, soit 1 020 minutes) : les athlètes ne peuvent pas finir bien au-delà.
 La variable est donc approximativement normale au centre, mais ne suit pas parfaitement une loi normale aux extrémités.
